```python
import networkx as nx
import matplotlib.pyplot as plt
import osmnx as ox
```

```python
o_park_address = "24 Willie Mays Plaza, San Francisco, CA 94107"
palace_address = "3601 Lyon St, San Francisco, CA 94123"

o_park_coords = ox.geocode(o_park_address)
palace_coords = ox.geocode(palace_address)

center = ((o_park_coords[0] + palace_coords[0]) / 2, (o_park_coords[1] + palace_coords[1]) / 2)
dist = max(ox.distance.great_circle(o_park_coords[0], o_park_coords[1], center[0], center[1]), ox.distance.great_circle(palace_coords[0], palace_coords[1], center[0], center[1]))
graph = ox.graph_from_point(center_point=center, dist=dist)

weighted_graph = ox.distance.add_edge_lengths(graph)

o_park_node = ox.distance.nearest_nodes(weighted_graph, o_park_coords[1], o_park_coords[0])
palace_node = ox.distance.nearest_nodes(weighted_graph, palace_coords[1], palace_coords[0])

def a_star_heuristic(u, v):
    u_coords = (weighted_graph.nodes[u]['y'], weighted_graph.nodes[u]['x'])
    v_coords = (weighted_graph.nodes[v]['y'], weighted_graph.nodes[v]['x'])
    return ox.distance.great_circle(u_coords[0], u_coords[1], v_coords[0], v_coords[1])

a_star_path = nx.astar_path(weighted_graph, o_park_node, palace_node, heuristic=a_star_heuristic, weight='length')
a_star_length = nx.astar_path_length(weighted_graph, o_park_node, palace_node, heuristic=a_star_heuristic, weight='length')

print(o_park_node)
print("A* path:", a_star_path)
print("A* path length:", a_star_length)
```

```python
dijkstra_path = nx.dijkstra_path(weighted_graph, o_park_node, palace_node, weight='length')
dijkstra_length = nx.dijkstra_path_length(weighted_graph, o_park_node, palace_node, weight='length')
print("Dijkstra path:", dijkstra_path)
print("Dijkstra path length:", dijkstra_length)
```
