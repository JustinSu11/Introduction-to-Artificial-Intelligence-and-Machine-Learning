## Part 2: EDA

```python
features = breast_cancer.data.columns
print(features)
```

```python
corr = breast_cancer.data.corrwith(breast_cancer.target).sort_values()
sns.barplot(x=corr.values, y=corr.index)
plt.show()
```

### 2.1 Shape, types, missing values, duplicates

```python
df = breast_cancer.frame  # features + 'target'
print(df.shape)
print(df.dtypes.value_counts())
print('missing values:', df.isnull().sum().sum())
print('duplicate rows:', df.duplicated().sum())
```

### 2.2 Class balance
`target`: 0 = malignant, 1 = benign

```python
counts = df['target'].value_counts().sort_index()
counts.index = breast_cancer.target_names
print(counts)
print((counts / len(df)).round(3))
sns.barplot(x=counts.index, y=counts.values)
plt.ylabel('count')
plt.title('Class balance')
plt.show()
```

### 2.3 Summary statistics

```python
print(breast_cancer.data.describe().T[['mean', 'std', 'min', 'max']].round(3))
```

### 2.4 Distributions by class
Four features most tied to the target (from the bar plot above).

```python
top = ['worst concave points', 'worst perimeter', 'mean concave points', 'worst radius']
labeled = df.copy()
labeled['diagnosis'] = labeled['target'].map(dict(enumerate(breast_cancer.target_names)))
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, col in zip(axes.ravel(), top):
    sns.boxplot(data=labeled, x='diagnosis', y=col, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

### 2.5 Correlation between features

```python
corr_matrix = breast_cancer.data.corr()
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, cmap='coolwarm', center=0)
plt.title('Feature correlation matrix')
plt.show()

# strongly correlated pairs (|r| > 0.9), each pair listed once
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape, dtype=bool), k=1))
pairs = upper.stack()
pairs = pairs[pairs.abs() > 0.9].sort_values(ascending=False)
print(len(pairs), 'pairs with |r| > 0.9')
print(pairs.head(10).round(3))
```

### 2.6 Pairplot of the top features

```python
sns.pairplot(labeled, vars=top, hue='diagnosis', corner=True, plot_kws={'s': 15})
plt.show()
```

### EDA takeaways
- **Clean data:** 569 rows, 30 numeric features, 0 missing values, 0 duplicates. No cleaning needed.
- **Mild imbalance:** 357 benign (62.7%) vs 212 malignant (37.3%). Not severe, but accuracy alone can hide misses on malignant cases, so also check recall.
- **Scale differs a lot:** `mean area` goes up to 2501, `mean smoothness` is about 0.1. Matters for linear models; tree models like XGBoost don't care.
- **Strongest signals:** `worst concave points` (r = -0.79), `worst perimeter` (-0.78), `mean concave points` (-0.78), `worst radius` (-0.78). Since benign = 1, negative means higher values go with **malignant**. The box plots show clear separation between classes.
- **Weak signals:** `smoothness error`, `mean fractal dimension`, `texture error`, `symmetry error` are near 0 correlation with the target.
- **Redundant features:** 21 feature pairs have |r| > 0.9 (e.g. `mean radius` and `mean perimeter`, r = 0.998), because radius, perimeter and area measure the same size. Trees handle this fine, but a linear model would struggle with the redundancy.
