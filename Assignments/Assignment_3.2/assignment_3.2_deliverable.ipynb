## Problem

You run a small factory that produces two items: **A** (\$40 profit per unit) and **B** (\$60 profit per unit).
Each product needs time on two machines. Goal: **maximize total weekly profit**
while staying within machine hours and meeting a minimum quota for B.

| | Machine 1 (hrs/unit) | Machine 2 (hrs/unit) | Profit/unit |
|---|---|---|---|
| **Item A** | 2 | 3 | \$40 |
| **Item B** | 4 | 2 | \$60 |
| **Hours available/week** | 100 | 80 | |

**Weekly machine hours**
- **Machine 1:** 100 hours available per week.
- **Machine 2:** 80 hours available per week.

**Extra rules**
- **Overtime:** run Machine 2 on Saturdays for **+20 hours**, at a cost of **\$500**.
- **Quota:** produce **at least 10 units** of item B per week.

**Decision variables**
- `a` = units of Product A (continuous)
- `b` = units of Product B (continuous)
- `o` = 1 if Machine 2 runs Saturday overtime, else 0 (binary)

## Questions
1. Write the objective function for total profit.
2. Write the constraints for this problem.
3. Solve with PuLP. What is the optimal combination of A and B? What is the maximum weekly profit? Should you run Machine 2 on overtime?

---
## 1) Objective function

**Maximize** Z =

_(40 * a + 60 * b - 500 * o)_

---
## 2) Constraints

- Machine 1: 2a + 4b <= 100
- Machine 2 (with overtime): 3a + 2b <= 80 + 20o
- Minimum B: b >= 10
- Variable types / bounds:
    - Continuous: a >= 0, b >= 0
    - Binary: o in {0, 1}

---
## 3) Solve with PuLP

```python
import pulp

problem = pulp.LpProblem("problem", pulp.LpMaximize)
```

**Variables**

```python
variables = problem.add_variable_dicts("variables", ["a", "b"], lowBound=0, cat="Continuous")
overtime_flag = problem.add_variable("overtime_flag", cat="Binary")
```

**Objective**

```python
problem.setObjective(40 * variables["a"] + 60 * variables["b"] - 500 * overtime_flag)
print(problem.objective)
```

**Constraints**

```python
problem.addConstraint(variables["b"] >= 10, "minimum_B")
problem.addConstraint(2 * variables["a"] + 4 * variables["b"] <= 100, "machine_1")
problem.addConstraint(3 * variables["a"] + 2 * variables["b"] <= 80 + 20 * overtime_flag, "machine_2")
print(problem)
```

**Solve**

```python
problem.solve()
pulp.LpStatus[problem.status]
```

**Results**

```python
print('a (A):', variables["a"].varValue)
print('b (B):', variables["b"].varValue)
print('overtime (overtime):', overtime_flag.varValue)
print('Profit:', pulp.value(problem.objective))
```

---
## Answers
- **Optimal mix:** A = 15 , B = 17.5
- **Max weekly profit:** \$1650
- **Run Machine 2 overtime?** No
